## 8. Constructor in Python

### Definition
The `__init__()` method initializes an object's attributes when an object is created.

Python automatically calls it during normal object creation.

### Syntax

```python
class ClassName:
    def __init__(self):
        pass
```

### Important Points
- `__init__()` is a special initialization method.
- It is called automatically when an object is created normally.
- `self` refers to the object being initialized.
- It is commonly used to initialize instance attributes.

Default Constructor

````python
class Student:

    # Constructor
    def __init__(self):
        print("Student object created")

# Create two objects
s1 = Student()
s2 = Student()
````

## Parameterized Constructor

### Definition
A parameterized constructor accepts additional arguments to initialize an object's attributes with specific values.

### Example
Different students can have different names and marks.

### Syntax

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
```

### Explanation
- `name` and `marks` receive values when an object is created.
- `self.name` and `self.marks` store those values in the current object.
- Each object can store its own data.


````python
class Student:

    # Parameterized constructor
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks

    # Display student details
    def display(self):
        print("Name:", self.name)
        print("Marks:", self.marks)

# Create objects with different values
s1 = Student("Harsh", 85)
s2 = Student("Rahul", 90)

s1.display()
print("----------")
s2.display()
````

## Constructor with Default Arguments

A constructor can use default parameter values so that an object can be created without providing every argument.

### Example

```python
class Student:
    def __init__(self, name="Unknown"):
        self.name = name
```

- `Student()` uses `"Unknown"`.
- `Student("Harsh")` uses `"Harsh"`.

Note: Python does not support defining multiple `__init__()` methods as traditional constructor overloading. Default arguments are one way to handle different initialization cases.

````python
class Student:

    def __init__(self, name="Unknown"):
        self.name = name

    def display(self):
        print("Name:", self.name)

s1 = Student()
s2 = Student("Harsh")

s1.display()
s2.display()
````

````python
# Create a Student class with a default name "Unknown".
class Student:
    def __init__(self, name="Unknown"):
        self.name = name

s1 = Student()
print(s1.name)
````

````python
# Create a Car class with a default brand "Not specified".
class Car:
    def __init__(self, brand="Not specified"):
        self.brand = brand

c1 = Car()
print(c1.brand)
````

````python
# Create a Student class with default name and marks.
class Student:
    def __init__(self, name="Unknown", marks=0):
        self.name = name
        self.marks = marks

s1 = Student()
s2 = Student("Amit", 80)

print(s1.name, s1.marks)
print(s2.name, s2.marks)
````

````python
# Create an Employee class with a default department "General".
class Employee:
    def __init__(self, name, department="General"):
        self.name = name
        self.department = department

e1 = Employee("Rahul")
e2 = Employee("Priya", "IT")

print(e1.name, e1.department)
print(e2.name, e2.department)
````

````python
# Create a Product class with name, price and default quantity 1. Display the total cost.
class Product:
    def __init__(self, name, price, quantity=1):
        self.name = name
        self.price = price
        self.quantity = quantity

    def display_total(self):
        print("Product:", self.name)
        print("Total Cost:", self.price * self.quantity)

p1 = Product("Notebook", 50)
p2 = Product("Pen", 10, 5)

p1.display_total()
p2.display_total()
````
