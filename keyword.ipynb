## 7. The self Keyword

### Definition
`self` refers to the current instance of a class.

It is used to access instance attributes and other instance methods.

### Important Points
1. `self` is the conventional name for the first parameter of an instance method.
2. Python passes the current object automatically when the method is called through that object.
3. `self` is not a Python keyword, but using this name is the standard convention.

### Example
When `s1.display()` is called, Python passes `s1` as the first argument to the method.

```python
class Student:

    def display(self):
        print("Current object:", self)

# Create an object
s1 = Student()

# Call the method
s1.display()

# Print the same object
print(s1)
```

```python
class Student:
    def show_name(self):
        print("My name is Harsh")

s1 = Student()
s1.show_name()
```

```python
class Person:
    def greet(self, name):
        print("Hello,", name)

p1 = Person()
p1.greet("Rahul")
```

```python
class Student:
    def show(self):
        print("Name:", self.name)

s1 = Student()
s1.name = "Amit"

s2 = Student()
s2.name = "Priya"

s1.show()
s2.show()
```
