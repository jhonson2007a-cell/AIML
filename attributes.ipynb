## 5. Attributes in Python

### Definition
Attributes are variables associated with a class or an object.

### Types of Attributes
1. Instance attributes: Store data for an individual object.
2. Class attributes: Defined inside the class and shared at the class level.

### Example
A student may have a name, age, and branch. Each student object can have different values.

```python
class Student:

    # Class attribute
    college = "SSIPMT"

# Creating objects
s1 = Student()
s2 = Student()

# Instance attributes
s1.name = "Harsh"
s1.age = 22

s2.name = "Rahul"
s2.age = 21

# Display details
print(s1.name, s1.age, s1.college)
print(s2.name, s2.age, s2.college)
```

```python
# Create three student objects. Each should have a different name and marks, while sharing the same class attribute college. Print all details.
class Student:
    college = "SSIPMT"

s1 = Student()
s1.name = "Amit"
s1.marks = 78

s2 = Student()
s2.name = "Priya"
s2.marks = 92

s3 = Student()
s3.name = "Rahul"
s3.marks = 85

for student in [s1, s2, s3]:
    print(student.name, student.marks, student.college)
```
