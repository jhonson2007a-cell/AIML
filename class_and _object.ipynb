# Python Object-Oriented Programming (OOP)

## 1. Introduction to OOP

### Definition
Object-Oriented Programming (OOP) is a programming approach based on classes and objects.

It helps us organize code, reuse code, and represent real-world entities in a program.

### Real-World Examples
- Student
- Car
- Employee
- Mobile Phone

### Advantages of OOP
1. Code reusability
2. Better code organization
3. Easy maintenance
4. Represents real-world entities

### Main Concepts of OOP
1. Class
2. Object
3. Attributes
4. Methods
5. Constructor


## 2. Procedural Programming vs OOP

### Procedural Programming
Procedural programming organizes a program using statements and functions.

### Object-Oriented Programming
OOP organizes related data and behavior using classes and objects.

### Example
A student has a name, roll number, and marks. We can store these details using variables or organize them in a Student class.

### Difference
- Procedural programming: Focuses on procedures and functions.
- OOP: Focuses on objects containing data and related behavior.

````python
# Procedural programming example

student_name = "Harsh"
roll_number = 101
marks = 85

# Display student information
print("Name:", student_name)
print("Roll Number:", roll_number)
print("Marks:", marks)
````

````python
# Marks in three subjects
python_marks = 85
sql_marks = 80
ai_marks = 90

# Calculate total and average
total = python_marks + sql_marks + ai_marks
average = total / 3

print("Total Marks:", total)
print("Average Marks:", average)
````

````python
# Marks in five subjects
marks1 = 75
marks2 = 68
marks3 = 82
marks4 = 55
marks5 = 70

# Store marks in a list
marks = [marks1, marks2, marks3, marks4, marks5]

# Calculate total and percentage
total = sum(marks)
percentage = total / 5

# Check result
if min(marks) >= 33 and percentage >= 40:
    result = "Pass"
else:
    result = "Fail"

# Display result
print("Total Marks:", total)
print("Percentage:", percentage, "%")
print("Result:", result)
````

## 3. Class in Python

### Definition
A class is a blueprint or template used to create objects.

It defines the attributes and methods that its objects can have.

### Syntax

```python
class ClassName:
    pass
```

### Explanation
- `class`: Keyword used to define a class.
- `ClassName`: Name of the class.
- `pass`: Placeholder used when no implementation is written yet.

### Example
Student is a class that can represent students in a college.

````python
# Creating an empty class

class Student:
    pass

# Display the class
print(Student)
````

## 4. Object in Python

### Definition
An object is an instance of a class.

A class is a blueprint, while an object is an individual instance created from that blueprint.

### Example
- Class: Student
- Object 1: s1
- Object 2: s2

Multiple objects can be created from the same class.

````python
# Create a class
class Student:
    pass

# Create two objects
s1 = Student()
s2 = Student()

# Check their types
print(type(s1))
print(type(s2))

# Check if both variables refer to the same object
print(s1 is s2)
````
