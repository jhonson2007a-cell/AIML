## 6. Methods in Python

### Definition
A method is a function defined inside a class.

Methods represent actions or behavior associated with objects.

### Syntax

```python
class ClassName:
    def method_name(self):
        # Statements
        pass
```

### Example
A Student class may have a `display()` method that prints a message or student details.

````python
class Student:

    # Defining an instance method
    def display(self):
        print("I am a student")

# Creating an object
s1 = Student()

# Calling the method
s1.display()
````

Practice Questions
1. Create a Calculator class with a welcome() method.
2. Create a Dog class with a bark() method.
3. Create a Calculator class with an add(a, b) method that prints the sum.
4. Create a Rectangle class with a method that calculates area using length and width.
5. Create a Student class with a method result(marks) that prints "Pass" if marks are at least 40; otherwise, print "Fail".

````python
# Create a Calculator class with a welcome() method.
class Calculator:
    def welcome(self):
        print("Welcome to Calculator")

c1 = Calculator()
c1.welcome()
````

````python
# Create a Calculator class with an add(a, b) method that prints the sum.
class Calculator:
    def add(self, a, b):
        print(a + b)

c1 = Calculator()
c1.add(10, 20)
````

````python
# Create a Rectangle class with a method that calculates area using length and width.
class Rectangle:
    def area(self, length, width):
        print("Area:", length * width)

r1 = Rectangle()
r1.area(5, 4)
````

````python
# Create a Dog class with a bark() method.
class Dog:
    def bark(self):
        print("Woof!")

d1 = Dog()
d1.bark()
````

````python
# Create a Student class with a method result(marks) that prints "Pass" if marks are at least 40; otherwise, print "Fail".
class Student:
    def result(self, marks):
        if marks >= 40:
            print("Pass")
        else:
            print("Fail")

s1 = Student()
s1.result(75)
s1.result(30)
````
